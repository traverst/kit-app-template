```python
import socket
import uuid
import msgpack
import sys
from confluent_kafka import Producer, Consumer, KafkaError, KafkaException

class KafkaHandler:
    
    running = False
    
    def __init__(self, kafka_host="localhost:9092", group_id="PythonGroup", auto_offset_reset="smallest"):
        self.kafka_host = kafka_host
        self.producer_conf = {
            'bootstrap.servers': kafka_host,
            'client.id': socket.gethostname()
        }
        self.consumer_conf = {
            'bootstrap.servers': kafka_host,
            'group.id': group_id,
            'auto.offset.reset': auto_offset_reset
        }
        self.producer = Producer(self.producer_conf)
        self.consumer = None  # Initialize as None; it will be set up in the consume_messages method

    def acked(self, err, msg):
        if err is not None:
            print(f"Failed to deliver message: {msg}: {err}")
        else:
            print(f"Message produced: {msg}")

    def send_message(self, topic, key_value, message_to_send):
        self.producer.produce(topic, key=key_value, value=msgpack.packb(message_to_send, use_bin_type=True)) 
        self.producer.poll(1)

    def send_query(self, topic, message):
        unique_id = uuid.uuid4().hex
        self.send_message(topic, unique_id, message)
        return unique_id

    def process_messages(self, key, message):
        print(f"Key: {key}")
        print(message)

    def stop_consuming(self):
        self.running = False

    def consume_messages(self, topic, process_fn=None):
        # Reinitialize the consumer each time consume_messages is called to avoid issues after closing it
        self.consumer = Consumer(self.consumer_conf)
        self.consumer.subscribe([topic])
        self.running = True
        print("starting up consumer")

        try:
            while self.running:
                msg = self.consumer.poll(timeout=1.0)
                if msg is None:
                    continue
                if msg.error():
                    if msg.error().code() == KafkaError._PARTITION_EOF:
                        print(f"%% {msg.topic()} [{msg.partition()}] reached end at offset {msg.offset()}")
                    else:
                        raise KafkaException(msg.error())
                else:
                    key = str(msg.key())
                    value = msgpack.unpackb(msg.value(), raw=False)
                    if process_fn:
                        process_fn(key, value)
                    else:
                        self.process_messages(key, value)
                sys.stdout.flush()  # <--- Flush print buffer

        except KeyboardInterrupt:
            print("Consumer interrupted. Exiting...")
        finally:
            self.consumer.close()
            print("Consumer Closed")
```

### LambdaStorage

This class stores the a function (by a generated uuid) and then later will execute at fuction (recalled by uuid) with the provided parameters.

```python
from typing import Any, Callable, Dict

class LambdaStorage:
    def __init__(self):
        self.function_store: Dict[str, Callable] = {}
    
    def store_function(self, *, uuid: str, func: Callable):
        """
        Store a function and return its UUID key
        """
        self.function_store[uuid] = func
    
    def get_function(self, *, function_id: str) -> Callable:
        """
        Retrieve a function by its UUID
        """
        return self.function_store.get(function_id)
    
    def remove_function(self, *, function_id: str) -> bool:
        """
        Remove a function from the store by its UUID
        
        Returns:
            bool: True if function was removed, False if it didn't exist
        """
        if function_id in self.function_store:
            del self.function_store[function_id]
            return True
        return False
    
    def execute_function(self, *, function_id: str, **kwargs) -> Any:
        """
        Execute a stored function with provided arguments
        """
        func = self.get_function(function_id = function_id) # retrieve the function matching the key.
        self.remove_function(function_id = function_id) # don't need it anymore.
        if func is None:
            raise KeyError(f"No function found with ID: {function_id}")
        return func(kwargs)
```

### datarequest


Stores the data as a group.  Consisting of the request type (the kafka topics... DatabaseQuery, DatabaseWrite)

### Kafka producer and consumer

Class that handles the kafka interface.

1. Build a handler: `kafka_handler = AsyncKafkaHandler(kafka_host="localhost:9092", group_id="PythonGroup")`
2. Start the producer: `await kafka_handler.start_producer()`
3. Start the consumer: `consumer_task = asyncio.create_task(kafka_handler.consume_messages(topic= "DatabaseResult"))`
4. Send the requests: `await kafka_handler.send_request(request= request)`

The request will have the details of the type of request, the datalog script and a response handler.

The lodging a request will see the request sent to the kafka topic and the response stored.

The consumer will receive the response and then call the response handler on the return data.


```python
import asyncio
import socket
import uuid
import msgpack
import sys
import nest_asyncio
from aiokafka import AIOKafkaProducer, AIOKafkaConsumer
from queue import Queue

# Apply nest_asyncio to allow nested event loops in Jupyter
nest_asyncio.apply()

class AsyncKafkaHandler:

    def __init__(self, kafka_host="localhost:9092", group_id="PythonGroup"):
        self.kafka_host = kafka_host
        self.producer = AIOKafkaProducer(
            bootstrap_servers=kafka_host,
            client_id=socket.gethostname()
        )
        self.consumer = None
        self.group_id = group_id
        self.message_queue = Queue()  # Queue to pass messages to the main thread

        self.messagehandlers = LambdaStorage()

    async def start_producer(self):
        await self.producer.start()

    async def stop_producer(self):
        await self.producer.stop()

    async def send_message(self, *, topic, key_value, message_to_send):
        key_bytes = key_value.encode('utf-8') if key_value else None
        await self.producer.send_and_wait(
            topic,
            key=key_bytes,
            value=msgpack.packb(message_to_send, use_bin_type=True)
        )

    async def send_request(self, *, request: datarequest):
        unique_id = uuid.uuid4().hex
        self.messagehandlers.store_function(uuid=unique_id, func=request.requesthandler)
        await self.send_message(topic=request.requesttype, key_value= unique_id, message_to_send= request.requestdata)
        return unique_id

    async def consume_messages(self, *, topic):
        self.consumer = AIOKafkaConsumer(
            topic,
            bootstrap_servers=self.kafka_host,
            group_id=self.group_id,
            auto_offset_reset='earliest'
        )
        await self.consumer.start()
        print("Starting up consumer")
        sys.stdout.flush()

        try:
            async for msg in self.consumer:
                key = str(msg.key.decode('utf-8')) if msg.key else None
                value = msgpack.unpackb(msg.value, raw=False)
                # Put the processed message into the message_queue
                self.message_queue.put((key, value))

        except Exception as e:
            print(f"Consumer error: {e}")
            sys.stdout.flush()
        finally:
            await self.consumer.stop()
            print("Consumer Closed")
            sys.stdout.flush()

    def stop_consume(self):
        self.consumer.stop()

    def run_queued_messages(self):
        """Retrieve messages from the queue."""
        messages = []
        while not self.message_queue.empty():
            key, value = (self.message_queue.get())
            messages.append(self.messagehandlers.execute_function(function_id= key, kwargs= value))
        return messages


```

```python
## Claude Sonat generated!

from typing import Dict, Any, List
import re
from decimal import Decimal

def parse_property_name(key: str) -> tuple[str, str]:
    """
    Parse a property key into group and name components.
    E.g., 'Material*Ambient.Blue' -> ('Material', 'Ambient.Blue')
    """
    if '*' in key:
        group, name = key.split('*', 1)
        return group, name
    return '', key

def detect_value_type(value: Any) -> str:
    """
    Determine the appropriate Datomic value type for a given Python value.
    Returns one of: 'string', 'int', 'float', 'bool', 'vector', 'entity'
    """
    if isinstance(value, bool):
        return 'bool'
    elif isinstance(value, int):
        return 'int'
    elif isinstance(value, (float, Decimal)):
        return 'float'
    elif isinstance(value, (list, tuple)) and len(value) == 3:
        # Assuming any 3-element sequence is a vector3
        return 'vector'
    elif isinstance(value, dict):
        return 'entity'
    else:
        return 'string'

def create_property_entity(name: str, value: Any, group: str = '') -> Dict:
    """
    Create a property entity with the appropriate value type.
    Includes group as a property if specified.
    """
    value_type = detect_value_type(value)
    
    property_entity = {
        ':property/name': name,
    }
    
    # Add group if present
    if group:
        property_entity[':property/group'] = str(group)
    
    # Add the appropriate value field based on type
    if value_type == 'string':
        property_entity[':property/string-value'] = str(value)
    elif value_type == 'int':
        property_entity[':property/int-value'] = int(value)
    elif value_type == 'float':
        property_entity[':property/float-value'] = float(value)
    elif value_type == 'bool':
        property_entity[':property/bool-value'] = bool(value)
    elif value_type == 'vector':
        property_entity[':property/vector-value'] = [float(x) for x in value]
    elif value_type == 'entity':
        # For nested objects, recursively convert them
        nested_entity = convert_to_datomic_entity(value)
        if nested_entity:  # Only include if the nested entity is not empty
            property_entity[':property/entity-value'] = nested_entity
        else:
            return None
    
    return property_entity

def convert_to_datomic_entity(data: Dict[str, Any]) -> Dict:
    """
    Convert a Python dictionary into a Datomic entity format.
    Properties maintain their group information as a property attribute.
    Returns None if the entity has no valid properties.
    """
    if not data:
        return None
        
    # Extract the name/identifier from the key
    key = next(iter(data))  # Get the first (and only) key
    properties_data = data[key]
    
    if not properties_data:  # Skip if no properties
        return None
    
    result = {
        ':object/name': key,
        ':object/properties': []
    }
    
    # Process all properties, maintaining group information
    for prop_key, value in properties_data.items():
        group, name = parse_property_name(prop_key)
        property_entity = create_property_entity(name, value, group)
        if property_entity:  # Only add if property creation was successful
            result[':object/properties'].append(property_entity)
    
    # Return None if no valid properties were found
    return result if result[':object/properties'] else None

def process_entities(data: Dict[str, Dict[str, Any]]) -> List[Dict]:
    """
    Process multiple entities from a dictionary of entity data.
    Skips empty entities and returns a list of valid Datomic entities.
    """
    datomic_entities = []
    
    for key, properties in data.items():
        entity_data = {key: properties}
        entity = convert_to_datomic_entity(entity_data)
        if entity:  # Only include non-empty entities
            datomic_entities.append(entity)
    
    return datomic_entities

```
