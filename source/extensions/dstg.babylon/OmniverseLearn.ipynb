## Omniverse extension development guide.

### Create New extension

Assuming you have installed omnikit

For windows:

` ./repo.bat template new `

select Extension, Python UI Extension and give it a name.

more at https://github.com/NVIDIA-Omniverse/kit-app-template

### Update file to setup for offline build

add the following to premake5.lua

 ```
 -- Include pip packages installed at build time
repo_build.prebuild_link {
    { "%{root}/_build/target-deps/pip_prebundle", ext.target_dir.."/pip_prebundle" },
} 
```

add the following to extension.toml

```
[[python.module]]
path = "pip_prebundle"
```

```

You will also need to edit to out of extension files to get the offline packaging working.

Firstly repo.toml in \kit-app-template needs to have the following added.

```
files = [
    ["**"],
    ["../../_build/target-deps/pip_prebundle/**"],   ## <-- this is what gets added!!
]
```

Secondly the pip.toml file in \kit-app-template\tools\deps will need all of the following

``` 
[[dependency]]
python = "../../_build/target-deps/python"  # Make sure this path is valid within your repo
packages = ["openpyxl==3.1.3","dill==0.3.9", "numpy==1.23","pandas==2.2.3", "sympy==1.13.3", "matplotlib", "scipy", "scikit-learn", "fastai"]
target = "../../_build/target-deps/pip_prebundle"  # Use a directory that will be included in the package
platforms = ["*"]
download_only = false
append_to_install_folder = true
python_include_dir = true
subprocess_path = ["${root}/_build/host-deps/msvc/MSBuild/Current/Bin", "${root}/_build/host-deps/msvc/VC/Tools/MSVC/14.29.30133/bin/HostX64/x64", "${root}/_build/host-deps/msvc/VC/Auxiliary/Build", "${root}/_build/host-deps/winsdk/bin/x64"]
msvc_path = "${root}/_build/host-deps/msvc/VC/Tools/MSVC/14.29.30133"
win_sdk_path = "${root}/_build/host-deps/winsdk"
```

you can add/remove packages to the packages line!!

lastly you will need to build the project to get it all working


Windows:
```
./repo.bat build
```

### Run the app 

Start up the app (start with usd composer)

```
./repo.bat launch
```

### Setup the Debugger

First step is to set up the debugger so that you can modify the code in VSCode and see the results reflected directly in Omniverse

 - In Omniverse app go to the windows menu -> extensions

 - Then select the VS Code Debugger Kit and activate it. (and make it autoload)

 - Dock the resulting window.

 - with the extension window select the "Open in VSCode" (to the right of the autoload option) 

    - You only need to do this once at setup

 - You will know VSCode is working when you hit refresh in the VS Code Debugger Attached is working
   -  If it isn't then you need to open (not the extension) but the kit-app-template an then run the debugger in VS Code (Python Remote attach)
      This should cause the Omniverse attach to change to connected!
   

#### Setup the Extension

Now you can load your extension via the Omniverse Extensions.

 - In Omniverse app go to the windows menu -> extensions
 
 - activate your extension and make it autoload

 - dock the resulting window

 - save the layout (layout menu)
